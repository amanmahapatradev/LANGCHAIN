# Google GenAI & LangChain Text Embeddings

This notebook demonstrates how to generate dense vector embeddings using Google GenAI with:
1. **LangChain (`GoogleGenerativeAIEmbeddings`)**
2. **The official Google GenAI SDK (`google.genai.Client`)**


```python
import os
from dotenv import load_dotenv
from langchain_google_genai import GoogleGenerativeAIEmbeddings

# 1. Load environment variables from .env file
load_dotenv()

# 2. Retrieve the API key by the variable name 'GOOGLE_API_KEY'
api_key = os.getenv("GOOGLE_API_KEY")

# 3. Initialize LangChain Google Embeddings with gemini-embedding-001
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001",
    google_api_key=api_key
)

# 4. Embed a sample query
query_text = "Retrieval-Augmented Generation with LangChain and Google GenAI"
vector = embeddings.embed_query(query_text)

print(f"Embedding successfully generated!")
print(f"Vector dimensions: {len(vector)}")
print(f"First 5 float values: {vector[:5]}")
```

```python
# Alternative: Using the official google-genai Client directly
from google import genai

client = genai.Client(api_key=api_key)

response = client.models.embed_content(
    model="gemini-embedding-001",
    contents="Retrieval-Augmented Generation with LangChain and Google GenAI"
)

sdk_vector = response.embeddings[0].values
print(f"SDK Vector dimensions: {len(sdk_vector)}")
print(f"First 5 float values: {sdk_vector[:5]}")
```
