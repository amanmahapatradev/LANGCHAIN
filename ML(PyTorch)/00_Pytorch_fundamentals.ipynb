Matrix Multiplication

Two main ways of performing multiplication in neural networks and deep learning:

1.Element-wise multiplication

2.Matrix multiplication (Dot product)

There are two main rules that performing matrix multiplicaion needs to satisfy:

1.The **inner dimentions** must match:

* (3,2) @ (3,2) Won't work
* (2,3) @ (3,2) will work
* (3,2) @ (2,3) will work

2.The resulting matrix has the shape of the outer dimentions:

* (2,3) @ (3,2) -> (2,2)
* (3,2) @ (2,3) -> (3,3)


```python
torch.matmul(torch.rand(3,10), torch.rand(10,3))
```

```python
torch.matmul(torch.rand(2,3), torch.rand(3,2))
```

### One of the most common error in deep learning: shape errors

```python
# Shapes for matrix multiplication
tensor_A = torch.tensor([[1,2],[3,4],[5,6]])
tensor_B = torch.tensor([[7,10],[8,11],[9,12]])
# torch.mm is same as torch.matmul ( it's an alias for writing less code )
torch.mm(tensor_A, tensor_B).shape 
```

To fix our tensor shape issues, we can manipulate the shape of one of our tensors using a **transpose** .

A **transpose** switches the axis or dimentions of a given tensor.

```python
tensor_B , tensor_B.shape
```

```python
tensor_B.T , tensor_B.T.shape
```

```python
# The matrix multiplication operation works when tensor_B is transposed
print(f"Origional shape: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}")
print(f"New shape: tensor_A = {tensor_A.shape} (same shape as above), tensor_B.T = {tensor_B.T.shape} ")
print(f"Multiplying: {tensor_A.shape} @ {tensor_B.T.shape} <- Inner dimention must match")
print("Output:\n")

output = torch.mm(tensor_A,tensor_B.T)

print(output)
print(f"\nOutput shape: {output.shape}")
```

## Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping -> reshapes an input tensor to a define shape 
* View -> Return view of an input tensor of certain shape but keep the same memory as the origional tensor
* Stacks -> Combine multiple tensor ont the top of eachother (vstack) or side by side (hstack)
* Squeeze -> remove all '1' dimentions from a tensor 
* Unsqueeze -> add a '1' dimenions to a target tensor
* Permute -> Return a view of the input with dimentions permuted (swapped) in a certain way

```python
# let's create a tensor 
import torch 
x = torch.arange(1.,10.)
x, x.shape
```

```python
# Add an extra dimention
x_reshaped = x.reshape(1,9)
x_reshaped , x_reshaped.shape 
```

```python
# Change the view
z = x.view(1,9)
z, z.shape
```

```python
# Changing z changing x ( because a view of a tensor shares the same memory as the origional input )
z[:, 0] = 5
z, x 
```

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x,x,x,x], dim=0)
x_stacked
```

```python
# torch.squeeze() -> removes all single dimentions from a target tensor
print(f"Previous tensor: {x_reshaped} ")
print(f"Previous shape: {x_reshaped.shape}")

# Remove extra dimentions from x_reshaped
x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed} ")
print(f"New shape: {x_squeezed.shape}")
```

```python
# torch.unsqueeze() -> adds a single dimention to a target tensor at a specific dim (dimention)
print(f"Previous target: {x_squeezed}")
print(f"Previous shape:{x_squeezed.shape}")

# add an extra dimention with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
# torch.permute -> rearrange the dimentions of a target tensor in a specified order
x_origional = torch.rand(size=(224,224,3)) # [height , width , color_channels]

# Permute the original tensor to rearrenge the axis (or dim) order
x_permuted = x_origional.permute(2, 0, 1) # shipts axis 0->1, 1->2, 2->0

print(f"Previous shape: {x_origional.shape}")
print(f"New shape: {x_permuted.shape}") # [color_channels , height , width]
```

### PyTorch tensors & NumPy

NumPy is a popular scientific Python numerical computing library.

And because of this, PyTorch has functionality to interact with it.
* Data is NumPy, want in PyTorch tensors -> torch.form_numpy(ndarray) 
* PyTorch tensor -> NumPy -> torch.Tensor.numpy()

```python
# NumPy array to tensor 
import torch 
import numpy as np 

array = np.arange(1.0,8.0)
tensor = torch.from_numpy(array) # Worning: When converting from numpy -> pytorch, It reflects numpy default datatype of float64 unless specified otherwise
array, tensor
```

```python
# Change the value of array, what will this do to 'tensor' ?
array = array + 1
array, tensor
```

```python
# Tensor to NumPy array 
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor,numpy_tensor 
```

```python
# Change the tensor, what happens to 'numpy_tensor'?
tensor = tensor + 1
tensor , numpy_tensor
```

## Reproducbility ( trying to take of random)

In short how a neural network learning:

'starting with random number -> tensor operations -> update random number to try and make them better representations of the data -> again -> again -> again .....'

To reduce the randomness in neural networks and PyTorch comes the concept of a random seed.

Essencialy what the random seed does is "flavour" the randomness.

```python
import torch 

# Create two random tensors
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Let's make some random but reproducible tensors
import torch 

# Set the random seed
RANDOM_SEED = 42 

torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3,4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3,4) 

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
