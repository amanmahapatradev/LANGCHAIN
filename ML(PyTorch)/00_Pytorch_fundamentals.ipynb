Matrix Multiplication

Two main ways of performing multiplication in neural networks and deep learning:

1.Element-wise multiplication

2.Matrix multiplication (Dot product)

There are two main rules that performing matrix multiplicaion needs to satisfy:

1.The **inner dimentions** must match:

* (3,2) @ (3,2) Won't work
* (2,3) @ (3,2) will work
* (3,2) @ (2,3) will work

2.The resulting matrix has the shape of the outer dimentions:

* (2,3) @ (3,2) -> (2,2)
* (3,2) @ (2,3) -> (3,3)


```python
torch.matmul(torch.rand(3,10), torch.rand(10,3))
```

```python
torch.matmul(torch.rand(2,3), torch.rand(3,2))
```

### One of the most common error in deep learning: shape errors

```python
# Shapes for matrix multiplication
tensor_A = torch.tensor([[1,2],[3,4],[5,6]])
tensor_B = torch.tensor([[7,10],[8,11],[9,12]])
# torch.mm is same as torch.matmul ( it's an alias for writing less code )
torch.mm(tensor_A, tensor_B).shape 
```

To fix our tensor shape issues, we can manipulate the shape of one of our tensors using a **transpose** .

A **transpose** switches the axis or dimentions of a given tensor.

```python
tensor_B , tensor_B.shape
```

```python
tensor_B.T , tensor_B.T.shape
```

```python
# The matrix multiplication operation works when tensor_B is transposed
print(f"Origional shape: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}")
print(f"New shape: tensor_A = {tensor_A.shape} (same shape as above), tensor_B.T = {tensor_B.T.shape} ")
print(f"Multiplying: {tensor_A.shape} @ {tensor_B.T.shape} <- Inner dimention must match")
print("Output:\n")

output = torch.mm(tensor_A,tensor_B.T)

print(output)
print(f"\nOutput shape: {output.shape}")
```

## Finding the min,max,mean,sum,etc (tesor aggregation)

```python
# Create a tensor
x = torch.arange(0,100,10)
x
```
