RANDOM TENSORS

Rendom tensors are important because the way many neural networks learn is that they start with tensors full of random numbers and then adjust those rendom numbers to better represent the data.


Start with random numbers -> Look at the data -> Update random numbers -> Look at data -> Update rendom numbers

```python
# Create a random tensor of size (3,4)
random_tensor = torch.rand(3,4)
random_tensor
```

```python
# Create a random tensor with similar shape to an image tensor
random_image_size_tensor = torch.rand(size = (3,224,224)) # height, width, color channels (R,G,B)   
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

Tensor datatypes

Note: Tenesor datatypes is one of the 3 big errors you'll run into with Pytorch & deep learning:

1. Tensors not right datatype
2. Tensors not right shape
3. Tensors not on the right device

```python
# Float 32 tensor
float_32_tensor = torch.tensor([3.0,6.0,8.0], 
                                dtype=torch.float16, # what datatype is the tensor (e.g. float32 or float16)
                                device= None,  # What device is your tensor on
                                requires_grad=False) # Whether or not to track gradients with this tensors operations
float_32_tensor
float_32_tensor.dtype
```

```python
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
float_16_tensor * float_32_tensor
```

```python
int_32_tensor = torch.tensor([3,6,9], dtype = torch.long)
int_32_tensor
```

```python
float_32_tensor * int_32_tensor
```

Getting information from tensor ( Tensor attributes )
1. Tensors not right datatype -> to get dtype from a tensor, can use [tensor.dtype]
2. Tensors not right shape -> to get shape from a tensor, can use [tensor.shape]
3. Tensors not on the right device -> to get device from a tensor, can use [tensor.device]

```python
# Create a tensor
some_tensor = torch.rand(3,4)
some_tensor
```

```python
# Filter out details about some tensors
print(some_tensor)
print(f" Datatype of tensor: {some_tensor.dtype}")
print(f" Shape of tensor: {some_tensor.shape}")
print(f" Data tensor is on: {some_tensor.device}")
```
